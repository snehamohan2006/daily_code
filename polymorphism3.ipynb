```python
class vehicle:
    def __init__(sel,brand,model):
        self.brand=brand
        self.model=model

    def move(self):
        print("move!")

class car(vehicle):
    pass

class boat(vehicle):
    def move(self):
        print("sail!")

class plane(vehicle):
    def move(self):
        print("fly!")

car1=car("bmw","tesla")
boat1=boat("sneha","fam")
plane1=plane("louda","airindia")

for x in (car1,boat1,plane1):
    print(x.brand)
    print(x.model)
    x.move()
```
