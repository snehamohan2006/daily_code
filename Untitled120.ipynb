```python
class node:
    def __init__(self,data):
        self.data=data
        self.next=None

first=node(10)
second=node(20)
third=node(40)

first.next=second
second.next=third

head=first

new_node=node(40)

current=head.next

new_node.next=current.next
current.next=new_node

current=head

while current:
    print(current.data,end="->")
    current=current.next

print("None")
```
