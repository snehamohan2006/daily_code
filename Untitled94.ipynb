```python
class node:
    def __init__(self,data):
        self.data=data
        self.next=None

first=node(10)
second=node(20)
third=node(30)
fourth=node(40)

first.next=second
second.next=third
third.next=fourth

head=first
print(head.data)
print(head.next.data)
print(head.next.next.data)
print(head.next.next.next.data)
```
