```python
class car:
    def __init__(self,car,brand):
        self.car=car
        self.brand=brand

    def move(self):
        print("drive")

class boat:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model

    def move(self):
        print("sail!")

class plane:
    def __init__(self,brand,model):
        self.brand=brand
        self.model=model

    def move(self):
        print("fly")

car1=car("bmw","tesla")
boat1=boat("sneha","fam")
plane1=plane("louda","airindia")

for x in (car1,boat1,plane1):
    print(x)
```
