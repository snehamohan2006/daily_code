```python
class node:
    def __init__(self,data):
        self.data=data
        self.next=None

first=node(10)
second=node(20)
third=node(30)
fourth=node(40)

first.next=second
second.next=third
third.next=fourth

head=first

search=30

current=head

while current:
    if current.data==search:
        print("found")
        break
    current=current.next
else:
    print("not found")
```
