```python
class node:
    def __init__(self,data):
        self.data=data
        self.next=None

one=node(10)
two=node(20)
three=node(30)

one.next=two
two.next=three

head=one

new_node=node(40)
current=head

while current:
    print(current.data)
    current=current.next

current.next=new_node
current=head

while current:
    print(current.data)
    current=current.next
```
