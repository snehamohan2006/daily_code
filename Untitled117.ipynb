```python
# Create a Node
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# Create nodes
first = Node(10)
second = Node(20)
third = Node(40)


# Connect the nodes
first.next = second
second.next = third


# Head points to the first node
head = first


# Create the new node
new_node = Node(30)


# Move current to node 20
current = head.next


# Insert 30 between 20 and 40
new_node.next = current.next
current.next = new_node


# Print the linked list
current = head

while current:
    print(current.data, end=" -> ")
    current = current.next

print("None")
```
